# Compute Brain Parechymal fraction

```python
def compute_bpf(id="00039", hunt=3):
    seg, _ = nfc.load_nifti(pil.get_segmentation_path_name(hunt, id, 3, absolute=True))
    bmask, _ = nfc.load_nifti(pil.get_brain_mask_path_name(hunt, id, pil.REGISTRATION_TO_MNI, absolute=True))
    return whole_brain_mc.compute_bpf(seg, bmask)
```

```python
matching_ids = pil.get_matching_ids()
bpf_np = np.zeros((len(matching_ids), 4))
for i, id in enumerate(matching_ids):
    bpf_hunt3 = compute_bpf(id, 3)
    bpf_hunt4 = compute_bpf(id, 4)
    difference = bpf_hunt3 - bpf_hunt4
    division = bpf_hunt4 / bpf_hunt3
    print(f"{id}:\tH3: {bpf_hunt3:.4f}\tH4: {bpf_hunt4:.4f}\tDec: {bpf_hunt3>bpf_hunt4}\tDiff: {difference:.4f}\tdiv: {division:.4f}")
    bpf_np[matching_ids.index(id)] = bpf_hunt3
```
