```python
import glob

path_imgs_h3 = "/media/agustin/LaCie/9410000000046/1"
path_imgs_h4 = "/media/agustin/LaCie2/anon9to22_hunt3-4/Forsendelse10_NT4MRI2 2018-11-16/9410000000046_1"

HUNT3_T1W_RAW = ['FAST_GEMS', 'EDR_GEMS']
HUNT3_T1W_BFC = ['FAST_GEMS', 'EDR_GEMS', 'FILTERED_GEMS']

HUNT4_T1W_RAW = ['EDR_GEMS', 'ACC_GEMS', 'IR_GEMS']
HUNT4_T1W_BFC = ['EDR_GEMS', 'FILTERED_GEMS', 'ACC_GEMS', 'IR_GEMS']



def same_elements(list1, list2):
    return set(list1) == set(list2)

def get_dicom_info(path):
    for folder in os.listdir(path):
        
        dcm_files = glob.glob(os.path.join(path, folder) + "/*.dcm")
   
        dicom = pydicom.dcmread(dcm_files[0], stop_before_pixels=True)
        dicom_attr = dicom["ScanOptions"]
        dicom_values = list(dicom_attr.value)

        if same_elements(HUNT3_T1W_RAW, dicom_values):
            print(f"HUNT3_T1W_RAW: {folder} {dicom_values}")
        elif same_elements(HUNT3_T1W_BFC, dicom_values):
            print(f"HUNT3_T1W_BFC: {folder} {dicom_values}")
        elif same_elements(HUNT4_T1W_RAW, dicom_values):
            print(f"HUNT4_T1W_RAW: {folder} {dicom_values}")
        elif same_elements(HUNT4_T1W_BFC, dicom_values):
            print(f"HUNT4_T1W_BFC: {folder} {dicom_values}")

        # print(f"{folder}: {dicom_values}")

print("HUNT4")
get_dicom_info(path_imgs_h4)
print("HUNT3")
get_dicom_info(path_imgs_h3)



```
