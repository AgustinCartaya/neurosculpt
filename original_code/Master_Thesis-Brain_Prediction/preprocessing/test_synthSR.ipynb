```python
def save_synthSR(img_in_path_name, img_out_path_name):
 
    command = ["mri_synthsr", 
               "--i", img_in_path_name, 
               "--o", img_out_path_name, 
               "--threads", "1"]
    
    result = subprocess.run(command, capture_output=True, text=True)

    if result.returncode == 0:
        # Print the output
        print(f"Freesurfer synthSR done for {img_in_path_name} ")
        # print(result.stdout)
    else:
        # Print error message
        print("Error:", result.stderr)
```
